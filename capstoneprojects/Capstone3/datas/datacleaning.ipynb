```python
#Change Runtime to Runtime (Min) so delete all the min in values
df['Runtime_min'] = df['Runtime'].str.replace(' min', '', regex=True)
df['Runtime_min'] = df['Runtime_min'].str.strip() #whitespace deletion
df['Runtime_min'].value_counts
```

```python
#Change Runtime_min to INT and drop Runtime
df['Runtime_min'] = df['Runtime_min'].astype(int)
df = df.drop('Runtime', axis=1)
df.info()
```

```python
df['Certificate'].unique()
#hybrid collection of ratings in India, America and Content Restricted (Unrated, Passed, Approved)
#Must add Content Restriction in agent (add new column for age categories)
```

```python
def mapping_certificate(cert):
    cert = str(cert).strip().upper()
    mapping = {'U': 'all ages',
               'G': 'all ages',
               'APPROVED': 'all ages',
               'PASSED': 'all ages',
               
               'UA': 'parental guidance',
               'U/A': 'parental guidance',
               'PG': 'parental guidance',
               'TV-PG': 'parental guidance',
               'GP': 'parental guidance',

               'PG-13': '13 years old and above',
               'TV-14': '14 years old and above',
               '16': '16 years old and above',
               
               'R': 'adults only',
               'A': 'adults only',
               'TV-MA': 'adults only',

               'UNRATED': 'not rated',
               'UNKNOWN': 'not rated'}
    
    return mapping.get(cert, 'Not rated')

df['age_group'] = df['Certificate'].apply(mapping_certificate)
df['age_group'].unique()
```

```python
#Check!
df.info()
```

```python
#bagi genrenya jadi list
df['Genre'] = df['Genre'].apply(lambda x: [genres.strip() for genres in x.split(', ')])
df['Genre']
```

```python
#make them lowercase
df['Genre'] = df['Genre'].apply(lambda x: [Gen.lower() for Gen in x])
df['Genre']
```

```python
#Trailing spaces sisa di Director, Stars, Series_Title
col = ['Director', 'Star1', 'Star2', 'Star3', 'Star4', 'Series_Title']
for columns in col:
    end_spaces = df[columns].str.endswith(' ').sum()
    start_spaces = df[columns].str.startswith(' ').sum()

    print(f'{columns} --> end: {end_spaces} ; start: {start_spaces}')
```
